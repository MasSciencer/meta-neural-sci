```python
import meep as mp
import numpy as np
import general_conf
import matplotlib.pyplot as plt
from scipy.integrate import quad
from scipy.interpolate import interp1d
from matplotlib.colors import Normalize
```

```python
wavelength = 3.7  # the wavelength of the source is 3.7 microns
M = 40 # the number of radial layers.
k = 2 * np.pi / wavelength # wave number
f = 5*1e3 # focal length 20 mm
```

```python
# Precompute radii
rt = np.sqrt(4 * f * (-np.pi / 2 + np.arange(1, M+1) * np.pi) / k)
```

```python
# Вычисление площадей кольцевых секторов
areas = []
for i in range(M-1):
    area = np.pi * (rt[i+1]**2 - rt[i]**2)  # Площадь кольца
    areas.append(area)

# Построение графика
plt.plot(np.arange(1, M), areas, marker='o', linestyle='-', color='b')
plt.xlabel('Порядковый номер сектора')
plt.ylabel('Площадь сектора')
plt.title('Зависимость площади сектора от его порядкового номера')
plt.grid(True)
plt.show()
```
