## Unites:
- Meep uses microns for every size instead of meters. So, pay attention!
- Wavelength should be calculated according notice below:  

**For example, suppose we are describing some photonic structure at infrared frequencies, where it is convenient to specify distances in microns. Thus, we let a
 = 1 μm. Then, if we want to specify a source corresponding to a vacuum wavelength of λ
 = 1.55 μm, we specify the frequency f
 as 1/1.55 = 0.6452. If we want to run our simulation for 100 periods, we then run it for 155 time units (= 100/f
). At a grid resolution (Δx
) of 10 pixels/μm and Courant factor (S
) of 0.5, one timestep is 0.05 time units or 0.16667 fs (Δt=SΔx/c)**

source: https://meep.readthedocs.io/en/master/Introduction/#units-in-meep

### Порядок действий

1. Последовательно вычисляем через meep конфигурации ячеек заданные уравнением. Получаем библиотеку зависимостей: форма ячейки - фазовый сдвиг + интенсивности первого порядка
2. Затем, на основе этих даных обучаем модель чтобы была построена ИИ модель зависимости [форма атома - параметры выходной волны]
3. Далее нужно как-то сделать сборку линзы из набора расчитанных мета-атомов
4. Каждая сборка проходит оценку на основе FOM при повторном расчете в meep
5. Модель обучается и оптимизируется для перебора ячеек и повышения FOM.
6. Самая эффективная сборка - финальный результат программы.
