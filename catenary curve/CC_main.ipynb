### Базовая модель распределения кривых и оптимизации

**Information:** 
    *Это самая базовая интерпритация моей программы, основання на оптимизации кривых, расположенных между радиусами для плоского фазового смещения. 
    А также с базовой интерпертацией функции catenary curve на cosh и коэффиценте кривизны А*
    
1. Радиусы соответствуют идеальным зонам Фурье
2. Коэффициенты кривой основаны на пред-тренированной модели
3. Цикл оптимизации будет подбирать коэффициенты

```python
# Настройка проекта
import numpy as np
import matplotlib.pyplot as plt
import meep as mp
import conf
import multiprocessing
import torch
from lib import sourses, numpy_functions
mp.verbosity(0) # decrease amount of output

"""
Преднастройки
TODO: перенести в конфиг
"""
aperture = conf.aperture
f = conf.f
cell_y = conf.cell_Y
cell_x = conf.cell_X
cell = mp.Vector3(cell_x, cell_y, 0)  # размер области симуляции
spot_ff = aperture
orders = 23 # the number of radial layers.
stack = 20 # Определяет количество саб-ячеек. При 0 имеет только одну главную ячейку
Q = 200 # curve detailing. It has almost no effect on the result
w = 0.5 # 1000 nm. The width of the gap between the curves
wavelength = conf.inc_wavelength
add_time = 4 
coef_resolution = 3
LIMIT_PRODUCT = 0.2 # 200 nm
angle = conf.angle
visualize_curve = True 
visualize_meep = True
```

```python
"""
Функции
"""
### Функции задания кривой

def setup_visualization():
    """Инициализация фигуры для визуализации"""
    global fig, ax
    fig, ax = plt.subplots(figsize=(10, 10))
    # Определяем палитру цветов
    global colors
    colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', 
          '#8c564b', '#e377c2', '#7f7f7f', '#bcbd22', '#17becf']

def plot_curve(xx, yy, X_sections, y0, n):
    """
    Функция для визуализации кривой и точек пересечения
    :param xx: x-координаты кривой
    :param yy: y-координаты кривой
    :param X_sections: точки пересечения с горизонтальной линией
    :param y0: y-координата горизонтальной линии
    :param n: номер кривой (для выбора цвета)
    """

    # Рисуем кривую
    ax.plot(xx, yy, color=colors[n % len(colors)], linewidth=0.5)

    # Выделяем точки пересечения красным
    ax.scatter(X_sections, np.full_like(X_sections, y0), color='red', s=10, zorder=3)
    

def calculate_radii (D, f, wavelength, orders):
    """
    Вычисляет радиусы на поверхности линзы, на которых происходит полный фазовый сдвиг от 0 до 2Pi.

    :param D: Диаметр линзы.
    :param f: Фокусное расстояние линзы.
    :param wavelength: Длина волны света.
    :param orders: Максимальное значение n для расчета радиусов.
    :return: radii: Список радиусов.
    """
    radii = []
    for n in range(1, orders + 1):
        r = np.sqrt((n * wavelength)**2 + 2 * n * wavelength * f)
        if r <= D / 2:  # Проверяем, чтобы радиус не превышал радиус линзы
            radii.append(r)
        else:
            break  # Прекращаем расчет, если радиус превышает размер линзы
    return radii

def catenary_function(x, x0, a):
    """
    Description:
    Используем самую простую функцию описания CC: со центровкой между двумя радиусами
    
    :param x: координата х
    :param x0: центр между радиусами
    :param a: коэффицент кривизны
    """   
    # https://chatgpt.com/share/67e139b7-8b94-800f-beef-0178c329f0d2
    
    return a * np.cosh((x - x0) / a)

def cut_y(xx, yy, y0):
    x_values = []
    for i in range(len(xx) - 1):
        y1, y2 = yy[i], yy[i + 1]
        if (y1 - y0) * (y2 - y0) <= 0:
            x1, x2 = xx[i], xx[i + 1]
            x_intersection = x1 + (y0 - y1) * (x2 - x1) / (y2 - y1)
            x_values.append(x_intersection)
    
    # Сортируем значения по возрастанию
    x_values.sort()

    return np.array(x_values)

def create_2d_curve(a_params, orders, stack, rt, Q=200, w=0.1, plot=False):
    """
    :param orders: количество порядков в линзе
    :param stack: количество кривых, которые пересекаются сечением на каждом порядке. Четное число 
    :param rt: лист с радиусами
    :param parameters: вектор или лист с коэффициентами формы
    :param Q: количество точек на кривую. Default 200 points
    :param w: The width of the gap between the curves. Default 100 nm
    :return: coordinates of X crossections points (as 2D numpy array) and centers
    """    
    list_pure_x = []
    y0 = -0.0001 + 2 * stack * w
       
    if plot:
        setup_visualization()
        ax.axhline(y=y0, color='black', linestyle='--', linewidth=1)
        
    for m in range(orders):
        if m == 0:
            rt0, x0 = 0, 0        
        else: 
            rt0 = rt[m - 1] + 0.01
            x0 = (rt[m] + rt0) / 2  # центры кривых
        
        a = a_params[m]
        x = np.linspace(rt0, rt[m] - 0.01, Q)
        y = catenary_function(x, x0, a)  # Предполагается, что `a` определено
        
        # Идём в обратном порядке по стеку (stack, stack-1, ..., 0)
        for n in range(stack, -1, -1):
            ip = np.argmin(y)
            y_shifted = y - y[ip]
            ymax = 0.5 * (np.max(x) - np.min(x))

            y_top = np.clip(y_shifted, -ymax, ymax)
            y_above_zero = np.clip(y_shifted - w, -ymax, ymax)

            xx = np.concatenate([x, x[::-1]])
            yy = np.concatenate([y_top, y_above_zero[::-1]]) + 2 * n * w
            
            x_pure = cut_y(xx, yy, y0)  # Предполагается, что `cut_y` определена
            list_pure_x.append(x_pure)
            
            if plot:
                # Визуализация
                plot_curve(xx, yy, x_pure, y0, n)
    
    # Преобразование в 2D numpy массив с размерностью (orders, stack + 1)
    x_pure = np.array(list_pure_x, dtype=object).reshape(orders, stack + 1)
    
    return x_pure

def calculate_width_and_centers(x_pure):
    """
    Вычисляет разницы в зависимости от количества элементов во вложенных массивах.
    
    Параметры:
    x_pure -- 2D numpy массив, где каждый элемент — массив с 1, 2 или 4 числами
    
    Возвращает:
    numpy массив с результатами (форма сохраняется, содержимое заменяется на разницы)
    """
    # Создаем выходной массив той же формы, но с dtype=object для гибкости
    result_width = np.empty_like(x_pure, dtype=object)
    result_centers = np.empty_like(x_pure, dtype=object)
    
    for i in range(x_pure.shape[0]):
        for j in range(x_pure.shape[1]):
            
            arr = x_pure[i][j]

            if len(arr) == 1 and i == 0 and j == 0:
                result_width[i][j] = np.array([2 * arr[0]]) # Пояснение, далее мы используем ширины которые центрированы. А 0 порядок это по сути урезанная ширина, так как другая сторона не учитывается. И чтобы ячейка не оказалась уменьшенной в 2 раза. Ее надо искусственно расширить.
                result_centers[i][j] = np.array([0])
                
            elif  len(arr) == 2:
                width1 = arr[1] - arr[0]
                result_width[i][j] = np.array([width1])
                center1 = arr[0] + width1 / 2
                result_centers[i][j] = np.array([center1])
                
            elif len(arr) == 4:
                width1 = arr[1] - arr[0]
                width2 = arr[3] - arr[2]
                result_width[i][j] = np.array([width1, width2])
                center1 = arr[0] + width1 / 2
                center2 = arr[2] + width2 / 2
                result_centers[i][j] = np.array([center1, center2])
                
            else:
                result_width[i][j] = np.array([0])
                result_centers[i][j] = np.array([0])
    
    return result_centers, result_width

#### Функции симмуляции

def get_width_and_center(shape_params, plot=True):
    """
    Description: Супер функция, которая возвращает numpy координаты центров и ширин для передачи в геометрию MEEP
    """
    rt = calculate_radii(aperture, f, wavelength, orders)
    x_pure = create_2d_curve(shape_params, orders, stack, rt, Q, w, plot=plot)
    centers, width = calculate_width_and_centers(x_pure)
    if plot:
        # Сбор данных для графиков
        stack_indices = list(range(width.shape[1]))
        orders_indices = list(range(width.shape[0]))
        width_values = [width[0][j][0] for j in stack_indices]
        center_values = [centers[j][0][0] for j in orders_indices]
        
        # Создаем фигуру с двумя subplots
        plt.figure(figsize=(12, 6))
        
        # График для ширины (левый subplot)
        plt.subplot(1, 2, 1)
        plt.plot(stack_indices, width_values, 'bo-', label=f'Width for {width.shape[1] - 1} stacks')
        plt.gca().xaxis.set_major_locator(plt.MaxNLocator(integer=True))
        plt.xlabel('Stack Index')
        plt.ylabel('Width Value')
        plt.title('Width vs Stack Index')
        plt.legend()
        plt.grid()
        
        # График для центров (правый subplot)
        plt.subplot(1, 2, 2)
        plt.plot(orders_indices, center_values, 'ro-', label=f'Centers for {width.shape[0]} orders')
        plt.gca().xaxis.set_major_locator(plt.MaxNLocator(integer=True))
        plt.xlabel('Order Index')
        plt.ylabel('Center Value')
        plt.title('Center vs Order Index')
        plt.legend()
        plt.grid()
        
        # Оптимизация расположения
        plt.tight_layout()
        plt.show()
        
    return centers, width

def infill_geometry(shape_params, geometry):
    centers, width = get_width_and_center(shape_params, visualize_curve)
    # add sub
    geometry.append(mp.Block(
        material=conf.in_material,
        size=mp.Vector3(conf.in_sub, mp.inf, 0),
        center=mp.Vector3(conf.sub_center.x),
    ))
    # add atoms
    for i in range(orders):
        for j in range(stack+1):
            for k in range (centers[i, j].size):
                if width[i][j][k] < LIMIT_PRODUCT: continue
                else:
                    ###---positive infilling---###
                    geometry.append(mp.Block(
                        material=conf.at_material,
                        size=mp.Vector3(x=conf.at_hei, y=width[i][j][k], z=0), 
                        center=mp.Vector3(x=conf.at_center.x, y=centers[i][j][k], z=0)
                    ))
    return geometry       

def setup_simulation(geometry, sources, k_point, plot=True, coef_resolution = coef_resolution):
    sim = mp.Simulation(
        cell_size=cell,
        boundary_layers= [mp.PML(thickness=conf.pml, direction=mp.X, side=mp.ALL)],
        geometry=geometry,
        sources=sources,
        k_point=k_point,
        resolution=conf.min_resolution * coef_resolution,
        symmetries = [mp.Mirror(mp.Y)]
    )
    
    # Adding a near2far area
    near2far_region = sim.add_near2far(
        conf.inc_fcen,  # центральная частота
        conf.inc_df,      # ширина спектра
        conf.inc_nfreq, # дискретность частот
        mp.Near2FarRegion(center=conf.flux_center, size=mp.Vector3(y=spot_ff)),
    )
    
    if plot:
        plt.figure(figsize=(1, 100))
        sim.plot2D(
            output_plane = mp.Volume(center=mp.Vector3(0,0,0), size=mp.Vector3(cell[0], cell[1],0))
        )
        plt.show()
    
    return sim, near2far_region

def MEEP_simulation(shape_params):
    """
    Definition: Включает в себя цикл программы: 
    1. заполнение линзы ячейками
    2. настройка и запуск симуляции
    3. интенсивность дальнего поля в двух сечениях
    4. расчет FOM based on multiplication of Strehl focal and deep 
    
    :param shape_params: вектор параметров а 
    :return: FOM based on Strehl
    """
    geometry = []
    geometry = infill_geometry(shape_params, geometry)
    sources_obj = sourses.KvectorSource(frequency=conf.inc_fcen, y = cell_y, angle=angle)
    sources = sources_obj.get_sources()
    k_point=sources_obj.get_k_vector()
    sim, near2far = setup_simulation(geometry, sources, k_point, visualize_meep)
    conf.run_end_time(sim, add_time)
    ff_res = conf.resolution_base/spot_ff
    Izyx_fp = calculation.get_Izyx(sim, near2far,ff_res, mp.Vector3(0, spot_ff, 0))  #Intensity in Focal Plane
    Izyx_oa = calculation.get_Izyx(sim, near2far,ff_res, mp.Vector3(spot_ff, 0, 0))  #Intensity in Optical Axis
    FOM = calculation.get_FOM(spot_ff, Izyx_fp, Izyx_oa)
    if visualize_meep:
        calculation.get_plot_FF(spot_ff, Izyx_fp)
        calculation.get_plot_FF(spot_ff, Izyx_oa)
        calculation.get_psf(conf.inc_nfreq, spot_ff, Izyx_fp)
    return FOM

shape_coefficient=torch.full((orders, 1), 10, requires_grad=False)
FOM = MEEP_simulation(shape_coefficient.detach().cpu().numpy())
```

```python
def run_single_simulation(args):
    """Функция для параллельного запуска симуляции"""
    shape_np, epsilon, delta = args
    perturbed_plus = shape_np + epsilon * delta
    perturbed_minus = shape_np - epsilon * delta

    new_FOM_plus = MEEP_simulation(perturbed_plus)
    new_FOM_minus = MEEP_simulation(perturbed_minus)

    grad = (new_FOM_plus - new_FOM_minus) / (2 * epsilon) * delta
    return grad

def compute_numerical_gradient_spsa_batch(shape_coefficient, epsilon=1, batch_size=5):
    """Параллельное вычисление градиента методом SPSA"""
    shape_np = shape_coefficient.detach().cpu().numpy()

    # Подготавливаем параметры для multiprocessing
    tasks = [(shape_np, epsilon, np.random.choice([-1, 1], size=shape_np.shape)) for _ in range(batch_size)]

    # Запускаем параллельные процессы
    with multiprocessing.Pool(processes=batch_size) as pool:
        gradients = pool.map(run_single_simulation, tasks)

    # Усредняем градиенты по батчу
    grad_avg = np.mean(gradients, axis=0)
    return torch.tensor(grad_avg, dtype=torch.float32)

def optimize_shape_coefficients(shape_coefficient,num_epochs=20, initial_lr=1, batch_size=5):
    """Оптимизация параметров с учётом параллельного градиентного вычисления"""
    momentum = torch.zeros_like(shape_coefficient)
    prev_FOM = -np.inf  
    history_fom = np.zeros(num_epochs)
    best_shape_coefficient = shape_coefficient
    best_FOM = 0

    for epoch in range(num_epochs):
        
        lr = initial_lr / (1 + 0.1 * epoch)  # Медленное убывание
        numerical_grad = compute_numerical_gradient_spsa_batch(shape_coefficient, batch_size=batch_size)
        momentum = 0.9 * momentum + lr * numerical_grad

        # Добавляем случайный шум
        momentum += 0.01 * torch.randn_like(numerical_grad)
        
        shape_coefficient = shape_coefficient + momentum

        FOM = MEEP_simulation(shape_coefficient.detach().cpu().numpy())
        
        if FOM > best_FOM:
            best_FOM = FOM
            best_shape_coefficient = shape_coefficient
        
        # Если FOM не улучшился несколько эпох – добавляем "отскок"
        if epoch > 5 and np.abs(FOM - prev_FOM) < 1e-5:
            shape_coefficient += 0.2 * (torch.rand_like(shape_coefficient) - 0.5)
        
        history_fom[epoch] = FOM
        print(f"Epoch {epoch+1}/{num_epochs} | FOM: {FOM:.6f} | lr: {lr:.2f}")

    return best_shape_coefficient, history_fom
```
